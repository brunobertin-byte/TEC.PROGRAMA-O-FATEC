```python
import numpy as np
import time
import timeit
from collections import Counter

def moda_devagar(lista_de_numeros):
  contagem ={}
  for i in lista_de_numeros:
    if i not in contagem:
      contagem[i] = 1
    else:
      contagem[i] += 1

  maior_ocorrencia = max(contagem.values())
  modas = []
  for valor, ocorrencia in contagem.items():
    if ocorrencia == maior_ocorrencia:
      modas.append(valor)
  return modas
```

```python
inteiros_aleatorios = np.random.randint(0, 100, size=1000)
inicio = time.time()
moda_devagar(inteiros_aleatorios)
fim = time.time()

print(f"tempo: {(fim - inicio):.4f} s")
```

```python
cron = timeit.Timer(
    stmt="moda_devagar(inteiros_aleatorios)",
    setup="from __main__ import moda_devagar, inteiros_aleatorios")

tempos = cron.repeat(repeat=5, number=1)
print(f"Melhor tempo: {min(tempos):.4f} s")
```

```python
def moda_counter(lista_de_numeros):
  contagem = Counter(lista_de_numeros)
  maior_ocorrencia = max(contagem.values())
  return [v for v, o in contagem.items() if o == maior_ocorrencia]
```

```python
cron = timeit.Timer(
    stmt="moda_counter(inteiros_aleatorios)",
    setup="from __main__ import moda_counter, inteiros_aleatorios")
print(f"Melhor tempo: {min(cron.repeat(repeat=5, number=1)): .4f} s")
```
